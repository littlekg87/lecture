## 1단계 — 텍스트 가져오기


````python
import re
import csv

# 1단계: TXT 파일 읽어오기
txt_path = 'practice_1.txt'  # 분석할 텍스트 파일 경로

with open(txt_path, 'r', encoding='utf-8') as file:
    lines = file.readlines()  # 파일의 모든 줄을 리스트로 저장

print('TXT 파일이 성공적으로 로드되었습니다!')
print(f'총 {len(lines)}개의 줄을 읽었습니다.')
````

---
## 2-1단계 — '구(口)' 문자가 포함된 줄 찾기


````python
# 2-1단계: '구(口)' 문자가 포함된 줄 찾기
gu_lines = []
for line in lines:
    if '口' in line:  # '口' 문자가 포함된 줄을 찾는다.
        gu_lines.append(line.strip())  # 줄 끝 공백 제거 후 저장

print(f"총 {len(gu_lines)}개의 '구(口)' 데이터가 발견되었습니다!")
````

---
## 2-2단계 — '구(口)'와 '。' 사이의 텍스트 추출

`re.search(r"口(.*?)。", line)` 은 **`口` 로 시작해서 `。` 로 끝나는 가장 짧은 구간**을 찾습니다.

- `(.*?)` — 사이에 있는 아무 글자나. `?` 가 붙으면 **최소한만** 가져옵니다.
- `match.group(1)` — 괄호 `( )` 안에 잡힌 부분, 즉 우리가 원하는 숫자입니다.


````python
# 2-2단계: '구(口)'와 '。' 사이의 텍스트 추출
gu_numbers = []
for line in gu_lines:
    match = re.search(r'口(.*?)。', line)  # '口'와 '。' 사이의 모든 텍스트 찾기
    if match:
        gu_numbers.append(match.group(1).strip())  # 찾은 텍스트만 리스트에 추가

print(f'총 {len(gu_numbers)}개의 데이터가 추출되었습니다!')
print(gu_numbers[:5])  # 앞부분만 출력해서 확인
````

---
## 3단계 — csv 형태로 저장하기


````python
# 3단계: CSV 파일로 저장
output_csv = 'sejong_gu_data_sample.csv'

with open(output_csv, mode='w', newline='', encoding='utf-8-sig') as file:
    writer = csv.writer(file)
    writer.writerow(['구(口) 데이터'])
    for row in gu_numbers:
        writer.writerow([row])

print(f'추출된 데이터가 {output_csv} 파일로 저장되었습니다!')
````

> 💡 `encoding='utf-8-sig'` 는 **엑셀에서 열었을 때 한글이 깨지지 않게** 해 줍니다.
> 그냥 `utf-8` 로 저장하면 윈도우 엑셀에서 글자가 깨져 보입니다.
